


Ideen er lidt at det skulle køre på onenote og være baseret på forløb og ikke noget der skal gøres hver dag. (Der står ikke noget i læreplanen om at det skal være hver gang)

Prøvede at lave en til svage elever og en til stærkere elever.

Simpel logbog (til svagere elever)

start på forløbet
1. Hvad skal jeg lave i Projektet

2. Hvad skal jeg lære i dette forløb?

3. Hvad ved jeg allerede om emnet

slut på forløbet

1. Hvad har jeg lavet i projektet?

2. Hvad har jeg lært?

3. Hvad fungerede godt i projektet?

4. Hvad kunne jeg gøre bedre næste gang?

5. Links til dokumentation


Udvidet logbog (til stærkere elever)

Start på forløbet
1. Problem / idé

Hvilket problem skal jeg løse?

2. Hvad skal jeg lære i dette forløb?

3. Plan for arbejdet

4. Mine egne læringsmål

Hvad vil jeg personligt blive bedre til?

Slut på forløbet
1. Hvad udviklede jeg?

Beskriv jeres it-system kort.

2. Hvad har jeg lært?

Beskriv hvilke faglige begreber eller metoder du har arbejdet med.

3. Eksempel på et problem jeg løste

Hvad var problemet?

Hvordan løste jeg det?

4. Evaluering af løsningen

Hvad fungerede godt?

Hvad kunne forbedres?

5. Hvis jeg havde mere tid

Hvad ville jeg videreudvikle?

6. Links til dokumentation
