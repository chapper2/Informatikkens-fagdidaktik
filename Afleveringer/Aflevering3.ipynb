2. Modul 90 minutter

Lav det således at spilleren mister et point hvis bolden rammer den påegældende spillers bagvæg.

Undersøg hvorledes boldens bevægelsesretning ændrer sig når den rammer en paddle og bagvæg. Hvad er forskellen og hvad er mest fysisk korrekt?

Implementer den fysisk mest korrekte metode

Det bliver for nemt at spille bolden til hinanden, undersøg hvordan du kan inmplementere en forhindring.
    - Brug din viden fra bevægelsesretningen til at implementere korrekt tilbagesendingsmetode på forhindringen

Ekstraopgaver
-Diskuter jeres mulige forbedringsforlag og implementer de af forslagene i stadig finder relevante.
