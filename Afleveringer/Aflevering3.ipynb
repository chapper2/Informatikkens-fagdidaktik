1. Modul 90 minutter
a) Læs om spillet pong
b) Åbn spillet og test spillet ved at trykke på det grønne flag
c)Diskuter med sidemakkeren hvordan spillet kan forbedres og lav en liste med mulige forbedringer. Husk at gemme listen til senere!

d) Forklar hvordan paddle virker for spilleren i højre side. Indsæt passende kommentarer til koden, der omhandler boldens og paddles bevægelse. Kommentarerne skal vise at i forstår hvad koden gør.

e) Lav en Spiller1 i venstre sider som virker på tilsvarende måde som den oprindelige
    Hjælp hvis du sidder fast: Saml de løse blokke og indsæt korrekt således du får en modstander i venstre side

f) Lav en vaiabel score som tæller et point op, hver gang spilleren rammer bolden

g) Udbyg det, således at hver spiller har deres individuelle score 
 

2. Modul 90 minutter

a) Lav det således at spilleren mister et point hvis bolden rammer den påegældende spillers bagvæg.

b) Undersøg hvorledes boldens bevægelsesretning ændrer sig når den rammer en paddle og bagvæg. Hvad er forskellen og hvad er mest fysisk korrekt?

c) Implementer den fysisk mest korrekte metode

d) Det bliver for nemt at spille bolden til hinanden, undersøg hvordan du kan inmplementere en forhindring.
    - Brug din viden fra bevægelsesretningen til at implementere korrekt tilbagesendingsmetode på forhindringen

Ekstraopgaver
-Diskuter jeres mulige forbedringsforlag og implementer de af forslagene i stadig finder relevante.
